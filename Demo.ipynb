#### How to Create a FAISS Index in RagLLM

The first thing to know is that there are 3 possible indices that you can create:
1. IndexFlatL2: does an exhaustive search resulting in high quality search results​
2. IndexIVFFlat: uses clustering to balance accuracy with search speed​
3. IndexPQ: uses compression to reduce memory usage and improve search speed

```python
from index_FlatL2 import IndexFlatL2
from index_IVFFlat import IndexIVFFlat
from index_PQ import IndexPQ
```
