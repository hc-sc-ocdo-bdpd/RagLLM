#### How to Create a FAISS Index

The first thing to know is that there are 3 possible indices that you can create:
1. IndexFlatL2 (`index_FlatL2.py`): does an exhaustive search resulting in high quality search results​
2. IndexIVFFlat (`index_IVFFlat.py`): uses clustering to balance accuracy with search speed​
3. IndexPQ (`index_PQ.py`): uses compression to reduce memory usage and improve search speed

Each of these Faiss files already has parameters pre-defined and inherits from a general `faiss_index.py` file which holds index functionality.

In addition to these four FAISS files, there's `split_folder.py` which splits a large folder of documents into smaller subfolders. This is useful when you want to embed and index documents in batches.

---

Let's go through a tutorial of how to create a FAISS index. For this example, we will create a FlatL2 index but the steps will be the same for any other type.

1. Create a .env file to store the OpenAI embedding model endpoint and API key. The file will literally be called `.env`. What you need to include in the file is written below. Just fill in the missing endpoint and API key.

```python
OPENAI_MODEL_ENDPOINT = "<INSERT MODEL ENDPOINT>"
OPENAI_API_KEY = "<INSERT API KEY>"
```

2. Install the requirements from `requirements.txt`. 

3. Import the class of the index you want to create.

```python
from index_FlatL2 import IndexFlatL2
```

4. Initialize your Flat index.

```python
flat = IndexFlatL2()
```

5. Each file you want to index must have an embedding created. You can either do this upfront and save the embeddings for later use, or create the embeddings at the same time as the index.

    Note: Files are appended to the index if it exists. The only way to start an index from scratch is to physically delete the folder containing the FAISS index and .pkl file.

- If you want to create the embeddings upfront, run the following code. `sample_files` is an example folder containing files to embed and index. The following code will create a folder (called `saved_files`) containing the embeddings, text chunks, and file metadata. Then, you can feed this folder into the `create()` method to create the folder (called `test_Flat`) containing the index and pickle file.

    Note: Make sure to save the `saved_files` folder if you plan to use it in the future because it will be written over every time `save_embeddings` is called.

```python
flat.save_embeddings("sample_files", True)
flat.create("saved_files", "test_Flat")
```

- If you want to create the embeddings at the same time as the index, run the following code. 

```python
flat.create("sample_files_2", "test_Flat")
```

6. Alternatively, if you have a Faiss index already built and want to add documents to it or query it, you can read it in.

```python
flat.read("test_Flat")
```

7. Lastly, if you want to query your Faiss index, run the following code. You can feed in a query and specify the number of results you want returned.

```python
query = "travel to australia"
flat.query(query, k = 3)
```
